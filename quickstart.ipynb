## Запрос

Замените `your_table` и `created_at` на нужные имена. Параметры `{start_date}` и `{end_date}` оставьте без кавычек.


````python
from parallel_request import extract

query = """
SELECT *
FROM your_table
WHERE created_at >= {start_date}
  AND created_at < {end_date}
"""
````

## DataFrame
